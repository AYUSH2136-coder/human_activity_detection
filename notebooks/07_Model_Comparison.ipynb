# 07 - Model Comparison

Comparing the performance of different models trained in previous notebooks.

```
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

plt.style.use("seaborn-v0_8-darkgrid")
```

```
# Assuming we collected accuracies manually from previous steps for simplicity
results = {
    "Model": ["Random Forest", "XGBoost", "1D CNN"],
    "Accuracy": [0.95, 0.97, 0.93]
}
df_results = pd.DataFrame(results)

plt.figure(figsize=(8, 5))
sns.barplot(data=df_results, x="Model", y="Accuracy")
plt.title("Model Comparison")
plt.ylim(0, 1.0)
for i, v in enumerate(df_results["Accuracy"]):
    plt.text(i, v + 0.02, str(v), ha="center")
plt.show()
```
