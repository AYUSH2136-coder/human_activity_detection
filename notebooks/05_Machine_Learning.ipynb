# 05 - Machine Learning

Training traditional ML models (Random Forest, XGBoost) on flattened windows.

```
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
import xgboost as xgb
from sklearn.metrics import classification_report, accuracy_score
import joblib
import os
```

```
X = np.load("../data/processed/X_windows.npy")
y = np.load("../data/processed/y_windows.npy")

# Flatten X for traditional ML
X_flat = X.reshape(X.shape[0], -1)
print("Flattened X shape:", X_flat.shape)
```

## Train/Test Split

```
X_train, X_test, y_train, y_test = train_test_split(X_flat, y, test_size=0.2, random_state=42, stratify=y)
```

## Random Forest Model

```
rf_model = RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1)
rf_model.fit(X_train, y_train)

rf_preds = rf_model.predict(X_test)
print("Random Forest Accuracy:", accuracy_score(y_test, rf_preds))
print(classification_report(y_test, rf_preds))
```

## XGBoost Model

```
# Adjust labels to start from 0 for XGBoost
min_label = y.min()
y_train_xgb = y_train - min_label
y_test_xgb = y_test - min_label

xgb_model = xgb.XGBClassifier(n_estimators=100, random_state=42, tree_method="hist")
xgb_model.fit(X_train, y_train_xgb)

xgb_preds = xgb_model.predict(X_test)
print("XGBoost Accuracy:", accuracy_score(y_test_xgb, xgb_preds))
print(classification_report(y_test_xgb, xgb_preds))
```

## Save Models

```
os.makedirs("../checkpoints/ml", exist_ok=True)
joblib.dump(rf_model, "../checkpoints/ml/random_forest.pkl")
xgb_model.save_model("../checkpoints/ml/xgboost.json")
print("Models saved.")
```
