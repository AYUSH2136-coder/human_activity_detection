# 08 - Inference Pipeline

Testing end-to-end inference using the trained XGBoost or Deep Learning model.

```
import numpy as np
import joblib
import xgboost as xgb
import os
```

## Load Model & Data

```
xgb_model = xgb.XGBClassifier()
xgb_model.load_model("../checkpoints/ml/xgboost.json")

X_test = np.load("../data/processed/X_windows.npy")[-100:]  # take last 100 windows as mock stream
# Flatten it
X_test_flat = X_test.reshape(X_test.shape[0], -1)
```

## Run Inference

```
predictions = xgb_model.predict(X_test_flat)
print("Predicted Classes for 100 samples:")
print(predictions)
```
