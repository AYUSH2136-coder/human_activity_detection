# 01 - Data Understanding

This notebook loads the raw MHEALTH dataset and explores the feature structure, activity labels, and subject distribution.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

plt.style.use("seaborn-v0_8-darkgrid")

DATA_PATH = Path("..") / "data" / "raw" / "mhealth_raw_data.csv"
SENSOR_COLUMNS = [
    "alx", "aly", "alz",
    "glx", "gly", "glz",
    "arx", "ary", "arz",
    "grx", "gry", "grz",
]
ACTIVITY_MAP = {
    0: "Null (no activity)",
    1: "Standing still",
    2: "Sitting and relaxing",
    3: "Lying down",
    4: "Walking",
    5: "Climbing stairs",
    6: "Waist bends forward",
    7: "Frontal elevation of arms",
    8: "Knees bending (crouching)",
    9: "Cycling",
    10: "Jogging",
    11: "Running",
    12: "Jump front & back",
}

df = pd.read_csv(DATA_PATH)

df["ActivityName"] = df["Activity"].map(ACTIVITY_MAP)
```

## Raw Dataset Overview

```python
print("Dataset path:", DATA_PATH)
print("Shape:", df.shape)
print("Columns:", list(df.columns))

df.head()
```

## Data Types and Missing Values

```python
df.info()
print("\nMissing values by column:")
print(df.isna().sum())
```

## Summary Statistics for Sensor Features

```python
df[SENSOR_COLUMNS].describe()
```

## Activity and Subject Distribution

```python
plt.figure(figsize=(12, 6))
sns.countplot(data=df, x="ActivityName", order=df["ActivityName"].value_counts().index)
plt.title("Activity Distribution")
plt.xticks(rotation=45, ha="right")
plt.ylabel("Count")
plt.tight_layout()
plt.show()

plt.figure(figsize=(10, 5))
sns.countplot(data=df, x="subject", palette="Set2")
plt.title("Number of Samples per Subject")
plt.xticks(rotation=45)
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```

## Basic Sensor Correlation

```python
plt.figure(figsize=(12, 10))
corr = df[SENSOR_COLUMNS].corr()
sns.heatmap(corr, cmap="coolwarm", center=0)
plt.title("Sensor Feature Correlation")
plt.tight_layout()
plt.show()
```

## Example Sensor Signal for Selected Activities

```python
selected_activity = 4  # Walking
selected_feature = "alx"
subset = df[df["Activity"] == selected_activity].head(500)

plt.figure(figsize=(12, 3))
plt.plot(subset.index, subset[selected_feature], label=selected_feature)
plt.title(f"{ACTIVITY_MAP[selected_activity]} - {selected_feature}")
plt.xlabel("Sample index")
plt.ylabel("Sensor value")
plt.legend()
plt.tight_layout()
plt.show()
```
